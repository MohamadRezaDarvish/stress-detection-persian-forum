## Interpretation rules for generated figures

The most important caveat is the before/after active-learning comparison. The historical and final project stages differ in label pools, split sizes, and leakage-control protocol. Therefore the change in MAE is a **descriptive project-stage comparison**, not a controlled causal estimate of the effect of active learning.

Likewise:

- a row-normalized confusion matrix visualizes recall behavior within each true class;
- SHAP/feature importance explains fitted-model behavior, not causality;
- test figures are frozen evidence and must not be used to justify new test-driven tuning.
